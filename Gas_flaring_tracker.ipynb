```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)

records = 300
start_date = datetime(2024, 1, 1)
companies = ['SPDC', 'NAOC', 'TotalEnergies', 'ExxonMobil', 'NDPR']
lgas = ['Bonny', 'Eleme', 'Gokana', 'Ikwerre', 'Ahoada West', 'Emeoha', 'Oyigbo']
flow_stations = [f'FS-{i:03d}' for i in range(1, 16)]

dates = [start_date + timedelta(days=random.randint(0, 500)) for _ in range(records)]
company_list = [random.choice(companies) for _ in range(records)]
lga_list = [random.choice(lgas) for _ in range(records)]
station_list = [random.choice(flow_stations) for _ in range(records)]

flared_mscf = np.random.uniform(500, 15000, records) 
utilization_pct = np.random.uniform(10, 85, records)

df = pd.DataFrame({
    'Flare_ID': [f'FLR-2024-{i:04d}' for i in range(1, records + 1)],
    'Date': dates,
    'Company': company_list,
    'LGA': lga_list,
    'Flow_Station': station_list,
    'Flared_Volume_Mscf': np.round(flared_mscf, 2),
    'Gas_Utilization_Pct': np.round(utilization_pct, 2)
})

df['Flared_Volume_MMscf'] = np.round(df['Flared_Volume_Mscf'] / 1000, 4)
df['CO2_Emissions_Tonnes'] = np.round((df['Flared_Volume_Mscf'] * 53.12) / 1000, 2)
df['Penalty_USD'] = np.round(df['Flared_Volume_Mscf'] * 2.00, 2)
df['Power_Potential_MWh'] = np.round(df['Flared_Volume_MMscf'] * 11.7, 2)
df['Households_Powerable'] = np.round((df['Power_Potential_MWh'] / 3) * 365, 0)
df = df.sort_values(by='Date').reset_index(drop=True)
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')

output_filename = 'Cleaned_Gas_Flaring_Data.csv'
df.to_csv(output_filename, index=False)

print(f"ETL Process Complete. Dataset saved to {output_filename}")

```
